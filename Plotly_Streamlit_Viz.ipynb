```python
import xarray as xr
import numpy as np
import plotly.graph_objects as go

# === Load NetCDF file ===
ds = xr.open_dataset("/content/drive/MyDrive/PyDataViz/Sample3D.nc")

# Extract coordinates
lon = ds["XAXIS"].values  # 90
lat = ds["YAXIS"].values  # 60
time = ds["TAXIS"].values  # 37

# Extract surface temperature (ZAX=0)
sst_surface = ds["TEMP"].isel(ZAX=0)  # shape = 90x60x37
sst_surface = sst_surface.where(sst_surface != -9999)  # remove missing values

# Create frames for animation
frames = []
for i, t in enumerate(time):
    z = sst_surface.isel(TAXIS=i).values  # transpose to match lat/lon orientation
    frame = go.Frame(
        data=[go.Surface(
            z=z,
            x=lon,
            y=lat,
            colorscale="Viridis",
            cmin=float(sst_surface.min()),
            cmax=float(sst_surface.max()),
            showscale=False
        )],
        name=str(i)
    )
    frames.append(frame)

# Initial surface (first timestep)
z0 = sst_surface.isel(TAXIS=0).values

fig = go.Figure(
    data=[go.Surface(
        z=z0,
        x=lon,
        y=lat,
        colorscale="Viridis",
        cmin=float(sst_surface.min()),
        cmax=float(sst_surface.max()),
        colorbar=dict(title="SST (°C)")
    )],
    layout=go.Layout(
        title="🌊 Animated Global Surface Temperature (SST)",
        width=900,
        height=700,
        template="plotly_dark",
        scene=dict(
            xaxis_title='Longitude',
            yaxis_title='Latitude',
            zaxis_title='SST (°C)',
            aspectratio=dict(x=2, y=1, z=0.3),
            camera=dict(eye=dict(x=1.5, y=1.5, z=0.8))  # initial view
        ),
        updatemenus=[dict(
            type="buttons",
            buttons=[
                dict(label="Play",
                     method="animate",
                     args=[None, {"frame": {"duration": 500, "redraw": True},
                                  "fromcurrent": True, "transition": {"duration": 0}}]),
                dict(label="Pause",
                     method="animate",
                     args=[[None], {"frame": {"duration": 0, "redraw": False},
                                    "mode": "immediate", "transition": {"duration": 0}}])
            ],
            x=0.1, y=0, xanchor="right", yanchor="top"
        )]
    ),
    frames=frames
)

fig.show()
```

```python
depth = ds["ZAX"].values      # 24 depth levels
# Select a fixed timestep (e.g., first time step)
t_index = 0
sst = ds["TEMP"].isel(TAXIS=t_index)

# Mask missing values
sst = sst.where(sst != -9999)

# Prepare frames for depth animation
frames = []
for i, z_val in enumerate(depth):
    z = sst.isel(ZAX=i).values  # transpose to match lat/lon
    frame = go.Frame(
        data=[go.Surface(
            z=z,
            x=lon,
            y=lat,
            colorscale='Viridis',
            cmin=float(sst.min()),
            cmax=float(sst.max()),
            showscale=False,
            hovertemplate='Lat: %{y}<br>Lon: %{x}<br>Temp: %{z:.2f} °C<extra></extra>'
        )],
        name=f"Depth: {z_val} m",
        layout=go.Layout(
            title_text=f"🌊 Temperature at Depth: {z_val} m, Time Step: {t_index+1}"
        )
    )
    frames.append(frame)

# Initial surface (shallowest depth)
z0 = sst.isel(ZAX=0).values

# Create figure
fig = go.Figure(
    data=[go.Surface(
        z=z0,
        x=lon,
        y=lat,
        colorscale='Viridis',
        cmin=float(sst.min()),
        cmax=float(sst.max()),
        colorbar=dict(title="Temp (°C)"),
        hovertemplate='Lat: %{y}<br>Lon: %{x}<br>Temp: %{z:.2f} °C<extra></extra>'
    )],
    layout=go.Layout(
        title=f"🌊 Temperature at Depth: {depth[0]} m, Time Step: {t_index+1}",
        width=900,
        height=700,
        template="plotly_dark",
        scene=dict(
            xaxis_title='Longitude',
            yaxis_title='Latitude',
            zaxis_title='Temperature (°C)',
            aspectratio=dict(x=2, y=1, z=0.3),
            camera=dict(eye=dict(x=1.5, y=1.5, z=0.8))
        ),
        updatemenus=[dict(
            type="buttons",
            buttons=[
                dict(label="Play",
                     method="animate",
                     args=[None, {"frame": {"duration": 400, "redraw": True},
                                  "fromcurrent": True, "transition": {"duration": 0}}]),
                dict(label="Pause",
                     method="animate",
                     args=[[None], {"frame": {"duration": 0, "redraw": False},
                                    "mode": "immediate",
                                    "transition": {"duration": 0}}])
            ],
            x=0.1, y=0, xanchor="right", yanchor="top"
        )]
    ),
    frames=frames
)

# Add depth slider
sliders = [dict(
    steps=[dict(method='animate',
                args=[[frame.name],
                      dict(mode='immediate',
                           frame=dict(duration=0, redraw=True),
                           transition=dict(duration=0))],
                label=f"{depth[i]} m") for i, frame in enumerate(frames)],
    transition=dict(duration=0),
    x=0, y=0,
    currentvalue=dict(font=dict(size=16), prefix="Depth: ", visible=True, xanchor='center'),
    len=1.0
)]

fig.update_layout(sliders=sliders)

fig.show()
```

```python
import xarray as xr
import numpy as np
import plotly.graph_objects as go

# === Load NetCDF file ===
ds = xr.open_dataset("/content/drive/MyDrive/PyDataViz/2020.nc")

# Replace with actual variable names
u = ds["u10"].isel(time=0)   # zonal wind
v = ds["v10"].isel(time=0)   # meridional wind
lat = ds["latitude"].values
lon = ds["longitude"].values

# === Subsample for clarity ===
step = 5
u_plot = u.values[::step, ::step]
v_plot = v.values[::step, ::step]
lat_plot = lat[::step]
lon_plot = lon[::step]

Lon, Lat = np.meshgrid(lon_plot, lat_plot)

# === Scale vectors for plotting ===
scale = 2  # adjust arrow length
Lon2 = Lon + u_plot * scale * 0.1
Lat2 = Lat + v_plot * scale * 0.1

# === Build interactive figure ===
fig = go.Figure()

# Add coastlines as base map
fig.update_geos(
    projection_type="natural earth",
    showcountries=True,
    showcoastlines=True,
    coastlinecolor="black",
    showland=True,
    landcolor="lightgray",
    showocean=True,
    oceancolor="lightblue",
)

# Add arrows as line segments
for i in range(Lon.shape[0]):
    for j in range(Lon.shape[1]):
        fig.add_trace(go.Scattergeo(
            lon=[Lon[i, j], Lon2[i, j]],
            lat=[Lat[i, j], Lat2[i, j]],
            mode="lines",
            line=dict(width=1, color="red"),
            showlegend=False
        ))

fig.update_layout(
    title="Interactive Wind Vectors (u10, v10) from 2020.nc",
    height=700,
    margin={"r":0,"t":30,"l":0,"b":0}
)

fig.show()
```

# Streamlit + Pandas Dashboard


```python
import streamlit as st
import pandas as pd
import numpy as np

# Page config
st.set_page_config(page_title='Basic Sales Dashboard', layout='wide')

# Generate sample data
np.random.seed(42)
df = pd.DataFrame({
    'Date': pd.date_range('2024-01-01', periods=100),
    'Sales': np.random.randint(500, 2000, size=100),
    'Region': np.random.choice(['North', 'South', 'East', 'West'], size=100),
    'Product': np.random.choice(['Product A', 'Product B', 'Product C'], size=100)
})

# Sidebar filters
st.sidebar.title('Filters')
regions = st.sidebar.multiselect('Select Region', df['Region'].unique(), default=df['Region'].unique())
products = st.sidebar.multiselect('Select Product', df['Product'].unique(), default=df['Product'].unique())

# Filter data
filtered_df = df[(df['Region'].isin(regions)) & (df['Product'].isin(products))]

# Display metrics
col1, col2, col3 = st.columns(3)
col1.metric("Total Sales", f"${filtered_df['Sales'].sum():,}")
col2.metric("Average Sales", f"${filtered_df['Sales'].mean():.0f}")
col3.metric("Records", len(filtered_df))

# Display filtered data
st.subheader("Filtered Data")
st.dataframe(filtered_df)
```
